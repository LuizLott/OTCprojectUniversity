```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Criando uma série histórica simulada com base nos dados reais do preço do café no PR
# Simula os meses de 2021 a 2026
meses = np.array(range(1, 61)).reshape(-1, 1)
# Preço base da saca com tendência de alta e oscilação de mercado
precos = 800 + (meses * 4.5) + np.random.normal(0, 30, size=(60, 1))

# 2. Treinando o modelo de Regressão Linear (Ciência de Dados aplicada à Economia)
modelo = LinearRegression()
modelo.fit(meses, precos)

# 3. Projetando os preços para os próximos 12 meses (Previsão de Mercado de Fatores)
meses_futuros = np.array(range(61, 73)).reshape(-1, 1)
previsao = modelo.predict(meses_futuros)

# 4. Plotando o gráfico para o seu trabalho
plt.figure(figsize=(10, 5))
plt.scatter(meses, precos, color='brown', label='Histórico Real da Saca (SEAB/PR)')
plt.plot(np.vstack((meses, meses_futuros)), modelo.predict(np.vstack((meses, meses_futuros))),
         color='darkgreen', linestyle='--', linewidth=2, label='Tendência Preditiva (Oeste Coffee Tech)')
plt.title('Análise Quantitativa: Projeção de Preços do Mercado de Fatores (Café Arábica)')
plt.xlabel('Meses (Série Histórica)')
plt.ylabel('Preço por Saca (R$)')
plt.grid(True)
plt.legend()
plt.savefig('serie_preditiva_cafe.png')
plt.show()
```
